```python
from sentence_transformers import SentenceTransformer

def embed(data):#https://huggingface.co/sentence-transformers
    model = SentenceTransformer('./models/all-MiniLM-L6-v2')#https://huggingface.co/sentence-transformers/all-MiniLM-L6-v2
    embeddings = model.encode(data).tolist()
    return embeddings

embeddings = embed("test string")
print(embeddings)
```
