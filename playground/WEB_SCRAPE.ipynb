```python
# Import required libraries

import requests

import lxml.html

from bs4 import BeautifulSoup

import re

# Define the base URL

base_url = 'https://www.fca.org.uk'

# Define the keyword to search for

keyword = 'CASS'

# Make a request to the sitemap and parse the response

response = requests.get('https://www.fca.org.uk/sitemap.xml')

soup = BeautifulSoup(response.text, 'lxml')

# Iterate over the sitemap entries

for entry in soup.find_all('url'):

    # Get the location and lastmod attributes

    location = entry.get('location')

    lastmod = entry.get('lastmod')

    

    # Check if the location contains the keyword

    if keyword in location:

        # Make a request to the location URL

        response = requests.get(f'{base_url}{location}')

        

        # Parse the HTML response

        soup = BeautifulSoup(response.text, 'lxml')

        

        # Extract the title and content

        title = soup.find('title').text

        content = soup.find('div', class_='container-block').get_text()

        

        # Print the results

        print('Title:', title)

        print('Content:', content)

        print('-------------------------------')
```
